```python
import math

class Solution:
    def maxPairStrength(self, num):
        i = 0
        j = i + 1
        maximum = 0

        while i < len(num) - 1:
            strength = (num[i] * num[j]) // (math.gcd(num[i], num[j]) ** 2)
            if strength > maximum:
                maximum = strength
            j += 1

            if j == len(num):
                i += 1
                j = i + 1

        return maximum
```
